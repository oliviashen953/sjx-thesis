```python
import os

traits = ["bmi", "brc", "height"]
base_directory = '/n/home00/jiaxinshen/NORMAL/'
num_groups = 20
num_iterations = 30

# Directory to store the generated job scripts
output_directory = "update_NORMAL_MCT_Jobs/"  # The directory where you want to save the .sh files
os.makedirs(output_directory, exist_ok=True)

# Define the Conda environment name
conda_env_name = 'my_env'

for trait in traits:
    for group_num in range(1, num_groups + 1):
        for iteration in range(1, num_iterations + 1):
            sh_file_name = f"job_{trait}_group{group_num}_iteration_{iteration}_MCT.sh"
            sh_file_path = os.path.join(output_directory, sh_file_name)
            
            with open(sh_file_path, 'w') as file:
                file.write("#!/bin/bash\n")
                file.write("#SBATCH -n 1\n")
                file.write("#SBATCH -N 1\n")
                file.write("#SBATCH -t 0-01:30:00\n")
                file.write("#SBATCH -p shared\n")
                file.write("#SBATCH --mem=40G\n")
                file.write("#SBATCH -c 2\n")
                file.write(f"#SBATCH -J job_{trait}_group{group_num}_iteration_{iteration}\n")
                
                file.write(f"#SBATCH -o {output_directory}/job_{trait}_group{group_num}_iteration_{iteration}_MCT.out\n")
                file.write(f"#SBATCH -e {output_directory}/job_{trait}_group{group_num}_iteration_{iteration}_MCT.err\n")
                file.write(f"#SBATCH --mail-user=YOUR_EMAIL\n\n")
                
                # Module load Mambaforge if needed
                file.write("module load Mambaforge/23.11.0-fasrc01\n")

                # Change directory to where the mc4 module is located
                file.write("cd /n/home00/jiaxinshen/mc4-tmp/MC4-1\n")

                # Activate conda environment
                file.write(f"conda activate {conda_env_name}\n\n")
                
                # Run Python script
                file.write(f"python mc4/command_line.py {base_directory}2.2_{trait}_ranked_transposed_files/ranked_transposed_{trait}_subgroup_{group_num}_iteration_{iteration}.csv -a mct -o column -hr 0\n")
          
                # Optionally deactivate the conda environment
                file.write("conda deactivate\n")

print("Shell scripts have been created in", output_directory)
```
